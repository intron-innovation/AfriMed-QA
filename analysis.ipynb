```python
for i in range(len(df)):
    if "answer is" in str(df['outputs'][i]).split(".")[0]:
        ans = str(df['outputs'][i]).split("answer is")[1].replace("(","").replace("*","").strip()[0]
    elif "option is" in str(df['outputs'][i]).split(".")[0]:
        ans = str(df['outputs'][i]).split("option is")[1].replace("*","").strip()[0]
    elif "correct options are" in str(df['outputs'][i]).split(".")[0]:
        ans = str(df['outputs'][i]).split("correct options are")[1].split("correct options are")[0].replace("*","").strip().replace("and ",",").replace("and",",").split(".")[0].replace(" ","").replace(",,",",")
    elif "correct answers are" in str(df['outputs'][i]).split(".")[0]:
        ans = str(df['outputs'][i]).split("correct answers are")[1].split("correct answers are")[0].replace("*","").strip().replace("and ",",").replace("and",",").split(".")[0].replace(" ","").replace(",,",",")
    else:
        ans = str(df['outputs'][i]).replace('**','').replace('*','').replace(' ','')[0]
    df['preds'][i] = ans
```

```python
for i in range(len(df)):
    if len(df['preds'][i]) > 1:
        cor = df['answer'][i] in df['preds'][i]
    else:
        cor = df["answer"][i] == df["preds"][i]
    
    df["correct2"][i] = cor
accuracy = df["correct2"].mean()
accuracy
```

```python
from google import genai
from google.genai import types
import base64

def generate():
    client = genai.Client(
        vertexai=True,
        project="afrimed-qa",
        location="global",
    )

    model = "gemini-2.5-pro-preview-05-06"
    contents = [
        types.Content(
            role="user",
            parts=[
                types.Part.from_text(text="""what is sentiments""")
            ]
        ),
    ]

    generate_content_config = types.GenerateContentConfig(
        temperature=1,
        top_p=1,
        seed=0,
        max_output_tokens=1024,
        safety_settings=[
            types.SafetySetting(
                category="HARM_CATEGORY_HATE_SPEECH",
                threshold="OFF"
            ),
            types.SafetySetting(
                category="HARM_CATEGORY_DANGEROUS_CONTENT",
                threshold="OFF"
            ),
            types.SafetySetting(
                category="HARM_CATEGORY_SEXUALLY_EXPLICIT",
                threshold="OFF"
            ),
            types.SafetySetting(
                category="HARM_CATEGORY_HARASSMENT",
                threshold="OFF"
            )
        ],
    )

    response_text = ""
    for chunk in client.models.generate_content_stream(
        model=model,
        contents=contents,
        config=generate_content_config,
    ):
        response_text += chunk.text

    return response_text

generate()
```

```python
import time
import pandas as pd
import vertexai
from vertexai.generative_models import GenerativeModel

model_name = "gemini-2.0-pro"
# Initialize Vertex AI
vertexai.init(project="afrimed-qa", location="us-central1")
model = GenerativeModel("gemini-2.5-pro-preview-05-06")

# Configurations
generation_config = {"max_output_tokens": 1024, "temperature": 1, "top_p": 0.95}

def generate_with_backoff(question, model, max_retries=5):
    retries = 0
    while retries < max_retries:
        try:
            responses = model.generate_content(
                [question], generation_config=generation_config, stream=True
            )
            return "".join(response.text for response in responses)
        except Exception as e:
            wait_time = 2 ** retries
            print(f"Error: {e}. Retrying in {wait_time} seconds...")
            time.sleep(wait_time)
            retries += 1
    return None  # Return None if all retries fail

```
